```python
# Transfer Learning
# Successful NLP Pretrained Models --------> HuggingFace (Free / Opensource)
# All Pretrained Models are prepared using Transformer Architectures
# Pretrained model for Sentimental Analysis


# Transformers internally have two parts
# 1. Encoder ---> used for classification tasks (Normal classification, sentiment analysis)
# 2. Decoder ---> Used for TEXT generation task

# Cycle / Flow of Transformers
#
# a. For Classification Task
#
# Text Input ----> Tokens ----> Embedding (To create Dense vector) + Positional Encoding ( remember the sequence of words in a given sentence)
#   -------> Stack on Encoder layer (Attention + ANN + Normalization) ----->Linear (dense) ----->Softmax ---> Classification Output
```

```python
#Step2: Import required libraries
from transformers import pipeline
```

```python
#Step3: Load a pretrained model (BERT based sentiment analysis pipeline)
# In this case it loads distillbert fine-tuned on SST2 (Stanford Sentiment Treebank)
nlp = pipeline("sentiment-analysis")
```

```python
nlp("I love this movie. Its absolutely fantastic!")
```

```python
nlp("This was the worst movie I've ever seen. Total waste of time.")
```

```python
print(f"Model used: {nlp.model.config._name_or_path}")
print(f"Tokenizer used: {nlp.tokenizer.name_or_path}")
```
