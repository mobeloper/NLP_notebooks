```python
# Check if Spacy is working properly

import spacy
model = spacy.load('en_core_web_sm')
```

```python
#POS Tagging
# POS is a process of NLP where each word in the given text data is assigned to the corresponding Parts of Speech
#
# NN - Noun
# VB - Verb
# JJ - Adjective
# RB - Adverb
# PRP - Pronoun
# IN - Preposition
# CC - Coordinating Conjunction
# DT - Determiner
#
# Why we need POS Tagging?
# POS Tagging helps the system to understand the GRAMATICAL STRUCTURE of the sentence.
# 1. Machine Translation
# 2. Text Analytics
# 3. IR (Information Retrieval)
```

```python
textExample = "Tesla is looking at buying U.S. Startups for $6 million"

data = model(textExample)

for record in data:
  print(record.text , record.pos_)
```

```python
#NER (Named Entity Recognition)
#
# NER is used in NLP apps to identofy and categorize ENTITIES in the text data into pre-defined categories
# a. Person Names
# b. Organizations
# c. Locations
# d. Dates
# e. Monetary Values
# f. Numercial Representation
# g. Symbols
# etc.

```

```python
model.get_pipe("ner").labels
```

```python

for record in data.ents:
  print(f" Token is : {record.text} , NER detected: {record.label_} ,Explain NER: {spacy.explain(record.label_)}")
```
