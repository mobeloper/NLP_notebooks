```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('.././Dataset/yelp.csv')
```

```python
data.info()
```

```python
data['sentiment'] = data['stars'].apply(lambda x: 'Happy' if x >= 3 else 'Sad')
```

```python
data.head()
```

```python
# Seperate  data  as features and label

features = data.iloc[:,[4]].values
label = data.iloc[:,10].values
```

```python
#Function to preprocess data
import string
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
def textPreprocessing(document):
  #Remove Punctuations
  processedData = ''.join([char for char in document if char not in string.punctuation])
  #Seperate words from document and normalize it
  wordsInLowerCase = [word.lower() for word in processedData.split(" ")]
  #Generate Vocab
  return [word for word in wordsInLowerCase if word not in stopwords.words('english')]
```

```python
# Create BOW

from sklearn.feature_extraction.text import CountVectorizer
wordVector = CountVectorizer(analyzer=textPreprocessing)
```

```python

#  Build the Vocab
finalWordVectorVocab = wordVector.fit(features)
```

```python

# Build the BOW
bagOfWords = finalWordVectorVocab.transform(features)
```

```python
# Create Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(bagOfWords,
                                                 label,
                                                 test_size=0.2,
                                                 random_state=6)
```

```python
#Build Model
from sklearn.linear_model import LogisticRegression
modelLogisticRegression = LogisticRegression()
modelLogisticRegression.fit(X_train,y_train)
```

```python
# Test whether model is good or bad

print("Train Score is {} and Test Score is {}".format(modelLogisticRegression.score(X_train,y_train), modelLogisticRegression.score(X_test,y_test)))
```

```python
#Build Model
from sklearn.neighbors import KNeighborsClassifier
modelKNeighborsClassifier = KNeighborsClassifier()
modelKNeighborsClassifier.fit(X_train,y_train)
```

```python
# Test whether model is good or bad

print("Train Score is {} and Test Score is {}".format(modelKNeighborsClassifier.score(X_train,y_train), modelKNeighborsClassifier.score(X_test,y_test)))
```

```python
#Build Model
from sklearn.svm import SVC
modelsvm = SVC()
modelsvm.fit(X_train,y_train)
```

```python

print("Train Score is {} and Test Score is {}".format(modelsvm.score(X_train,y_train), modelsvm.score(X_test,y_test)))
```

```python
#Build Model
from sklearn.tree import DecisionTreeClassifier
modeldt = DecisionTreeClassifier()
modeldt.fit(X_train,y_train)
```

```python

print("Train Score is {} and Test Score is {}".format(modeldt.score(X_train,y_train), modeldt.score(X_test,y_test)))
```

```python
#Build Model
from sklearn.ensemble import RandomForestClassifier
modelrf = RandomForestClassifier()
modelrf.fit(X_train,y_train)

print("Train Score is {} and Test Score is {}".format(modelrf.score(X_train,y_train), modelrf.score(X_test,y_test)))
```
