```python
#Load the dataset
from sklearn.datasets import fetch_20newsgroups
newsgroups = fetch_20newsgroups(remove=('headers','footers','quotes'))
docs_raw = newsgroups.data
print(len(docs_raw))
```

```python
# Ref: https://scikit-learn.org/stable/datasets/real_world.html
```

```python
docs_raw
```

```python
#Create Document Term Matrix | BOW
from sklearn.feature_extraction.text import CountVectorizer

tf_vectorizer = CountVectorizer(stop_words='english', #Stopwords
                               lowercase=True, # Normalization
                               token_pattern= r'\b[a-zA-Z]{3,}', #Tokenization - All words must be greater than equal to 3 a-zAZ
                                strip_accents='unicode' #Trims all unicode chars which is not a part of eng language
                               )

tfObject = tf_vectorizer.fit_transform(docs_raw)
```

```python
print(tfObject.shape)
```

```python
#Create TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

tfidfVectorizer = TfidfVectorizer()
tfidfObject = tfidfVectorizer.fit_transform(docs_raw)
```

```python
print(tfidfObject.shape)
```

```python
#Apply LDA to create topic clusters for Bag of Words

from sklearn.decomposition import LatentDirichletAllocation
lda_tf = LatentDirichletAllocation(n_components=20 , random_state=0)
lda_tf.fit(tfObject) # as per books passing BOW
```

```python
from sklearn.decomposition import LatentDirichletAllocation
lda_tfIDF = LatentDirichletAllocation(n_components=10 , random_state=0)
lda_tfIDF.fit(tfidfObject) # as per seasoned DS considering memory optimizn reqd for handling Bigdata
```

```python
#Visualizing the LDA model
import pyLDAvis
import pyLDAvis.lda_model
pyLDAvis.enable_notebook()
```

```python
#---------------------LDAModel,BOW/TFIDF Object,CountVectorizerObject ----------

pyLDAvis.lda_model.prepare(lda_tf,tfObject,tf_vectorizer)
```

```python
pyLDAvis.lda_model.prepare(lda_tfIDF,tfidfObject,tfidfVectorizer)
```

```python
# Assign label to each text data

import pandas as pd
```

```python
#Transform all document using the trained model
topicData = lda_tfIDF.transform(tfidfObject)
```

```python
#Select the topic with highest frequncy

labels = topicData.argmax(axis=1)
```

```python
finalDF = pd.DataFrame({
    'data': docs_raw,
    'label' : labels
})
```
