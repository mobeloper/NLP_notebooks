```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
dataset = [
    'I love NLP',
    'I love NLU',
    'Do you love NLP?',
    'NLU is amazing and awesome!'
]
```

```python
#OOV ---> Out Of Vocabulary
#
tokenizer = Tokenizer(oov_token="<DontKnow>")
```

```python
tokenizer.fit_on_texts(dataset)
```

```python
tokenizer.word_index
```

```python
sequence = tokenizer.texts_to_sequences(dataset)
```

```python
sequence
```

```python
paddedData = pad_sequences(sequence, padding='post')
paddedData
```

```python
testData = ['Welcome to my NLP class','My name is Prashant Nair','I am your trainer',"Welcome to my class on NLP and NLU"]
seq = tokenizer.texts_to_sequences(testData)
```
