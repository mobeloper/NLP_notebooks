```python
# Stemming and Lemmatization
# 1. Convert the given data into words (Apply Word Tokenization)
# 2. Apply Stemmer or Lemma
```

```python
# Stemming is a process of REDUCING a word into its base/root form.
# Stemming is a rule-based approach - It may or may not produce a VALID WORD
```

```python
textData = "running runners run easily fairly cache caching cached"
```

```python
words = word_tokenize(textData)
words
```

```python
#Apply Stemmer
import nltk
stemmer = nltk.PorterStemmer()
stemWords = [ stemmer.stem(word) for word in words ]
stemWords
```

```python
# Lemmatization: Its a more sophisticated process that reduces the word into its base
# for considering the base word is a language dictionary word
```

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmaWords = [ lemmatizer.lemmatize(word , pos='v') for word in words] #Return verb value
lemmaWords
```

```python
# WordNet --- Lexical Database
# Used for Semantic Understanding (getting some meaning out of the input)
# - Synonyms
# - Antonyms
# - definitions
# - Relationship between words

# WordNet --- Applicable for English Language
# Application
# 1. Word Similarity Search
# 2. Word Sense Disambiguation
# 3. Synonyms Extraction
```

```python
from nltk.corpus import wordnet
syn = wordnet.synsets("bank")
print("Definition of bank: ",syn[1].definition())
print("Examples of word: ",syn[1].examples())
```

```python
# Stopwords
# Stopwords are words that have little to NO SEMANTIC meaning

#Note: Usually Stopwords are removed for NLP-based Activities.
#      Stopwords are preserved for NLU based activities
```

```python
import nltk
nltk.download('stopwords')
```

```python
#Stopwords removal using NLTK

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
textData = "Welcome to my NLP class. I am Prashant Nair and I will teach you NLP"

words = word_tokenize(textData)

print("Before Stopwords Removal: ")
print(words)

#Filter Stopwords

print("After Stopwords Removal: ")

stopWordsList  = stopwords.words('english')
filteredWords = [ word for word in words if word not in stopWordsList ]
print(filteredWords)
```

```python
#Spacy Demo on Tokenization, Lemma, Stopwords

import spacy
model = spacy.load("en_core_web_sm")
```

```python
text = "The banks of the river were crowded. She deposited cash in the bank"
```

```python
#Process the text
documents = model(text)
```

```python
# 1. Sentence Tokenization

for sent in documents.sents:
  print(sent)
```

```python
# 2. Word Tokenization

for word in documents:
  print(word.text)
```

```python
# Stemming and Lemma (Spacy doesnt support Stemming)

for word in documents:
  print(f"{word.text}  --- {word.lemma_}")
```
