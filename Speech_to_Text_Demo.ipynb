```python
import speech_recognition as sr
```

```python
#Initialize the Recognizer
r = sr.Recognizer()
```

```python
#Load the data (WAV file)
dataFile = sr.AudioFile('./Datasets/Recording.wav')
```

```python
#Convert File data into Feature Data
#Recog -->record

with dataFile as source:
    audio = r.record(source)
```

```python
audio
```

```python
#Recognize the speech and get the sentence
#Models . google Recognition engine(local)

text = r.recognize_google(audio)
text
```

```python
#Getting Data from Live Mic
```

```python
with sr.Microphone() as source:
    print("Speak ...")
    a1 = r.listen(source)
```

```python
textspeech = r.recognize_google(a1)
textspeech
```

```python
#Store the speech text in file
with open('speechtext.txt', 'w') as fw:
    fw.write(textspeech)
    fw.close()
```
